Aqui se crean las configuraciones que representaran las superficies de Riemann de género 2, entonces empezamos por los ei que definen las ramificaciones de la superficie sobre la esfera de Riemann

```python
import numpy as np
import pandas as pd
from numpy.polynomial import Polynomial
```

```python
SEED=20260905
rng = np.random.default_rng(SEED)    #este seria solo el generador de numeros
min_gap = 0.08
```

```python
def branch_points(rng, min_gap):
    while True:
        lambdas = np.sort(rng.uniform(0,1,3))    #limites, tamaño 
        e= np.concatenate([[0],lambdas,[1],[2]])

        gaps= np.diff(e)

        if np.any(gaps < min_gap):
            continue
        else:
            break

    return e
```

```python
def polynomial_coefficients(e):
    y2=Polynomial.fromroots(e)
    return y2.coef
```

```python
polynomial_coefficients(np.array([0,0.2,0.5,0.9,1,2]))
```

```python
filas = []
for i in range(500):
    ramas = branch_points(rng, min_gap)
    coefs = polynomial_coefficients(ramas)

    fila = {
        "sample_id": f"S{i+1:05d}",
        "family_id": "interior",
        "seed": SEED,
    }
    for j, e in enumerate(ramas, start=1):
        fila[f"e{j}"] = e
    for j, c in enumerate(coefs):
        fila[f"c{j}"] = c
    fila["valid"] = True

    filas.append(fila)
```

```python
df = pd.DataFrame(filas)
df
```
