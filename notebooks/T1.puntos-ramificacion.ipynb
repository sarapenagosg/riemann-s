Aqui se crean las configuraciones que representaran las superficies de Riemann de género 2, entonces empezamos por los ei que definen las ramificaciones de la superficie sobre la esfera de Riemann

```python
import numpy as np
from numpy.polynomial import Polynomial
```

```python
rng = np.random.default_rng(20260905)    #este seria solo el generador de numeros
min_gap = 0.08
```

```python
def branch_points(rng, min_gap):
    while True:
        lambdas = np.sort(rng.uniform(0,1,3))    #limites, tamaño 
        e= np.concatenate([[0],lambdas,[1],[2]])

        gaps= np.diff(e)

        if np.any(gaps < min_gap):
            continue
        else:
            break

    return e
```

```python
def polynomial_coefficients(e):
    y2=Polynomial.fromroots(e)
    print("y^2 = ", y2.coef)
```
